```python
import requests
import pprint
from getpass import getpass

# APIキーを指定
api_key = getpass('APIキーを入力してください: ')

# 検索駅名を変数stationに格納import
station = input('駅名を入力してください: ')
```

```python
geocord_url = 'https://maps.googleapis.com/maps/api/geocode/json'
# データを取得
response = requests.get(
    geocord_url,
    params={
        'address': station,
        'key': api_key
    }
)




# レスポンスのJSONを取得
json_data = response.json()

# JSONの中身を確認
pprint.pprint(json_data)
```

```python

location = json_data['results'][0]['geometry']['location']
lat = location["lat"]
lng = location["lng"]

print(f'緯度{lat},経度{lng}')

```

```python
places_url = "https://places.googleapis.com/v1/places:searchNearby"


places_params = {
    "location": f"{lat},{lng}",
    "radius": 500,
    "includedTypes": "restaurant",
    "language": "ja",
    "key": api_key
}

places_response = requests.get(places_url, params=places_params)
places_json = places_response.json()

pprint.pprint(places_json)
```

```python


places_url = "https://places.googleapis.com/v1/places:searchNearby"

places_params = {
    "includedTypes": ["restaurant"],
    "maxResultCount": 20,
    "languageCode": "ja",
    "locationRestriction": {
        "circle": {
            "center": {
                "latitude": lat,
                "longitude": lng
            },
            "radius": 500
        }
    }
}

headers = {
    "Content-Type": "application/json",
    "X-Goog-Api-Key": api_key,
    "X-Goog-FieldMask": "places.displayName,places.formattedAddress,places.location"
}

places_response = requests.post(
    places_url,
    headers=headers,
    json=places_params
)

places_json = places_response.json()
pprint.pprint(places_json)
```

```python
for place in places_json.get("places"):
    name = place.get("displayName", {}).get("text", "不明")
    rating = place.get("rating", "評価なし")
    address = place.get("formattedAddress", "住所なし")

    print(f"レストラン名：{name}")
    print(f"評価：{rating}")
    print(f"住所：{address}")
    print("-" * 30)
```
